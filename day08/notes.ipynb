```python
def calculate_love_score(name1, name2):
    word1 = "True"
    word2 = "Love"
    true_score = 0
    love_score = 0

    for letter in name1:
        if letter in word1.lower():
            true_score +=1
        elif letter in word2.lower():
            love_score +=1
    
    for letter in name2:
        if letter in word1.lower():
            true_score +=1
        elif letter in word2.lower():
            love_score +=1


    print(str(true_score)+str(love_score))

calculate_love_score("Kanye West", "Kim Kardashian")
```
